```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\Girish\Downloads\customer_churn_sample.csv")

print("Dataset loaded successfully!")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

df.head()
```

```python
print("=== DATASET SHAPE ===")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n=== COLUMN NAMES ===")
print(df.columns.tolist())

print("\n=== DATA TYPES ===")
print(df.dtypes)

print("\n=== DATASET INFORMATION ===")
df.info()
```

```python
print("=== MISSING VALUES ===")

missing_values = df.isnull().sum()

print(missing_values)

print("\nTotal missing values:",
      df.isnull().sum().sum())


print("\n=== DUPLICATE ROWS ===")

duplicate_rows = df.duplicated().sum()

print("Total duplicate rows:", duplicate_rows)
```

```python
print("=== CATEGORICAL VALUE CHECK ===")

categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Fix the CustomerID name to a clean format
df = df.rename(columns={"customerid": "customer_id"})

print("=== STANDARDIZED COLUMN NAMES ===")
print(df.columns.tolist())
```

```python
df = df.rename(columns={
    "tenuremonths": "tenure_months",
    "subscriptiontype": "subscription_type",
    "monthlycharges": "monthly_charges",
    "totalcharges": "total_charges",
    "contracttype": "contract_type",
    "supporttickets": "support_tickets",
    "paymentmethod": "payment_method"
})

print("=== FINAL COLUMN NAMES ===")
print(df.columns.tolist())
```

```python
numeric_columns = [
    "age",
    "tenure_months",
    "monthly_charges",
    "total_charges",
    "support_tickets"
]

print("=== NUMERICAL DATA TYPE VALIDATION ===")

for col in numeric_columns:
    print(f"{col}: {df[col].dtype}")
```

```python
# Validate Customer ID structure

print("=== CUSTOMER ID VALIDATION ===")

# Check for missing Customer IDs
missing_ids = df["customer_id"].isnull().sum()
print("Missing Customer IDs:", missing_ids)

# Check for duplicate Customer IDs
duplicate_ids = df["customer_id"].duplicated().sum()
print("Duplicate Customer IDs:", duplicate_ids)

# Check Customer ID format
id_pattern = r"^CUST-\d{4}$"

invalid_ids = ~df["customer_id"].str.match(id_pattern, na=False)

print("Invalid Customer ID formats:", invalid_ids.sum())
```

```python
# Clean text and categorical columns

text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Text values cleaned successfully.")
```

```python
print("========================================")
print("       FINAL DATASET VALIDATION")
print("========================================")

print("\nDataset shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum())

print("\nTotal missing values:",
      df.isnull().sum().sum())

print("\nDuplicate rows:",
      df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# Export the cleaned dataset

output_file = "cleaned_customer_churn.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset exported successfully!")
print("File name:", output_file)
```
